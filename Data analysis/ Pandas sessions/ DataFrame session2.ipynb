#### Merge

```python
emp=pd.DataFrame({'id':[101, 102, 103, 104], 'name':['Akshay', 'Rohit', 'Rahul', 'Vedant'], 'did':[1, 2, 1, 4]})
emp
```

```python
dept = pd.DataFrame({'dept_id':[1, 2, 3], 'dname':['Admin', 'IT', 'Sales']})
dept
```

```python
pd.merge(emp, dept, how='inner', left_on='did', right_on='dept_id')
```

```python
pd.merge(emp, dept, how='left', left_on='did', right_on='dept_id')
```

```python
pd.merge(emp, dept, how='right', left_on='did', right_on='dept_id')
```

```python
pd.merge(emp, dept, left_on='did', right_on='dept_id')
```

```python
pd.merge(emp, dept, how='outer', left_on='did', right_on='dept_id')
```
