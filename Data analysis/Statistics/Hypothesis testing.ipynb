```python
import numpy as np
from scipy import stats
```

```python
#sample data
data = [70, 60, 74, 65, 78, 80, 64]
```

```python
#null hypothesis
mu = 70
```

```python
#t-test: sample size < 30
#z-test: sample size >= 30
#chi_test: sample is categorical
```

```python
statistics, p_value = stats.ttest_1samp(data, 70)
```

```python
p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print('Reject the null hypothesis')
else:
    print('Failed to reject null hypothesis')
    
```
