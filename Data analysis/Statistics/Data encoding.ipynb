```python
import numpy as np
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = np.array([['Low'], ['Medium'],['High'], ['Low'], ['High']])
data
```

```python
OE = OrdinalEncoder(categories=[['Low', 'Medium', 'High']], dtype=int)
```

```python
OE.fit_transform(data)
```

#### One Hot Encoding

OneHotEncoder class

```python
data = np.array([['Apple'], ['Orange'], ['Mango']])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
OHE = OneHotEncoder(sparse_output=False, dtype=int)
```

```python
OHE.fit_transform(data)
```
