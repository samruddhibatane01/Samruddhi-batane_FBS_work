```python
import matplotlib.pyplot as plt
```

#### Lineplot
1. Trends over time
2. Time series analysis
3. Upward/downward trend

```python
import numpy as np
```

```python
overs = np.arange(1, 21)
team_1 = np.random.randint(0, 37, 20)
team_2 = np.random.randint(0, 37, 20)
```

```python
team_1
```

```python
team1_score = np.cumsum(team_1)
team1_score
```

```python
team2_score = np.cumsum(team_2)
```

```python
plt.figure(figsize=(7,3))
plt.style.use('dark_background')
plt.plot(overs, team1_score, marker='o', color='red', label='team1')
plt.plot(overs, team2_score, marker='<', color='blue', label='team2')
plt.xlabel('Overs')
plt.ylabel('Score')
plt.title('Match summary')
plt.legend()
plt.show()
```

#### Scatterplot
1. Relationship between two variables
2. Correlation
3. Positive correlation
4. Negative correlation

```python
plt.figure(figsize=(7,3))
plt.scatter(overs, team_1, color='red',marker='o', label='team1')
plt.scatter(overs, team_2, color='blue', marker='<',label='team2')
plt.legend()
plt.show()
```

#### Pie chart
1. Showing percentage distribution
2. part to whole comparision
3. best to use when categories <=5

```python
marks = [70, 90, 67, 40, 76, 72]
subjects = ['m1', 'm2', 'm3', 'm4', 'm5', 'm6']
```

```python
plt.pie(marks, labels=subjects, autopct='%0.2f%%')
plt.show()
```
